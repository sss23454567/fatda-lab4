```python
import kagglehub
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import milp, LinearConstraint, Bounds
DATA = kagglehub.dataset_download("shkarupylomaxim/adtech-lab4-tv-audiences/versions/2")
programs = pd.read_parquet(f"{DATA}/programs.parquet")
viewings = pd.read_parquet(f"{DATA}/viewings.parquet")
daily_weights = pd.read_parquet(f"{DATA}/daily_weights.parquet")
viewer_attributes = pd.read_parquet(f"{DATA}/viewer_attributes.parquet")
attributes = pd.read_parquet(f"{DATA}/attributes.parquet")
rate_card = pd.read_parquet(f"{DATA}/rate_card.parquet")
survey = pd.read_csv(f"{DATA}/purchase_survey.txt", sep="\t", dtype=str, keep_default_na=False)
survey_layout = pd.read_csv(f"{DATA}/purchase_survey_layout.txt", sep="\t", dtype=str)
acr_sessions = pd.read_parquet(f"{DATA}/acr_sessions.parquet")
hh_attributes = pd.read_parquet(f"{DATA}/hh_attributes.parquet")
reference_ratings = pd.read_parquet(f"{DATA}/reference_ratings.parquet")
universe = pd.read_parquet(f"{DATA}/universe.parquet")
```

```python
# ## Task 1.0 — Clean the survey file
# %%

raw = survey.copy()
survey_clean = survey.copy()

bad_records = []

def report(kind, mask, action):
    bad_records.append({
        "Kind of bad record": kind,
        "Rows": int(mask.sum()),
        "Action": action
    })

mask = survey_clean.ne(
    survey_clean.apply(lambda x: x.str.strip())
).any(axis=1)

report("Extra spaces", mask, "Strip spaces")

survey_clean = survey_clean.apply(lambda x: x.str.strip())

mask = survey_clean["KEYWORD"] != survey_clean["KEYWORD"].str.upper()

report("Incorrect keyword case", mask, "Convert to uppercase")

survey_clean["KEYWORD"] = survey_clean["KEYWORD"].str.upper()

mask = survey_clean["RESPID"].str.startswith("0")

report("Leading zeros", mask, "Remove leading zeros")

survey_clean["RESPID"] = survey_clean["RESPID"].str.lstrip("0")

mask = ~survey_clean["RESPID"].str.fullmatch(r"\d+")

report("Invalid RESPID", mask, "Remove rows")

survey_clean = survey_clean[~mask].copy()

valid_keywords = [
    "X710100",
    "X710101",
    "X710102",
    "X710103",
    "X710201",
    "X710202",
    "X710203"
]

mask = ~survey_clean["KEYWORD"].isin(valid_keywords)

report("Unknown keyword", mask, "Remove rows")

survey_clean = survey_clean[~mask].copy()

mask = ~survey_clean["VALUE"].isin(["0", "1"])

report("Invalid VALUE", mask, "Remove rows")

survey_clean = survey_clean[~mask].copy()

mask = survey_clean["VALUE"] == "0"

report("Negative answers", mask, "Exclude from positive answers")

survey_clean = survey_clean[~mask].copy()

mask = survey_clean.duplicated(
    subset=["RESPID", "KEYWORD", "VALUE"]
)

report("Duplicate answers", mask, "Remove duplicates")

survey_clean = survey_clean[~mask].copy()

ages = viewer_attributes[
    viewer_attributes["attr_id"] == 300
].copy()

ages["value"] = pd.to_numeric(
    ages["value"],
    errors="coerce"
)

adult_ids = set(
    ages.loc[ages["value"] >= 18, "respondent_id"]
    .astype(str)
    .str.lstrip("0")
)

panel_ids = set(
    viewer_attributes.loc[
        viewer_attributes["attr_id"] == 101,
        "respondent_id"
    ].astype(str).str.lstrip("0")
)

adult_panel = adult_ids & panel_ids

mask = ~survey_clean["RESPID"].isin(adult_panel)

report("Not adult panel member", mask, "Remove rows")

survey_clean = survey_clean[~mask].copy()

pair_keywords = [
    "X710101",
    "X710102",
    "X710103"
]

pair_counts = (
    survey_clean[
        survey_clean["KEYWORD"].isin(pair_keywords)
    ]
    .groupby("RESPID")["KEYWORD"]
    .nunique()
)

two_pair_counts = pair_counts[pair_counts == 2]

conflicting_ids = pair_counts[pair_counts > 1].index

mask = (
    survey_clean["RESPID"].isin(conflicting_ids)
    & survey_clean["KEYWORD"].isin(pair_keywords)
)

report(
    "Conflicting pair counts",
    mask,
    "Keep highest pair category provisionally"
)

survey_clean["VALUE"] = survey_clean["VALUE"].astype(int)

survey_wide = survey_clean.pivot_table(
    index="RESPID",
    columns="KEYWORD",
    values="VALUE",
    aggfunc="max",
    fill_value=0
)

survey_wide = survey_wide.reindex(
    index=sorted(adult_panel),
    columns=valid_keywords,
    fill_value=0
)

survey_wide = survey_wide.rename(columns={
    "X710100": "buyer",
    "X710101": "pairs_1",
    "X710102": "pairs_2",
    "X710103": "pairs_3plus",
    "X710201": "STRIDE",
    "X710202": "KESTREL",
    "X710203": "EASYWALK"
})

survey_wide["pairs"] = np.select(
    [
        survey_wide["pairs_3plus"] == 1,
        survey_wide["pairs_2"] == 1,
        survey_wide["pairs_1"] == 1
    ],
    ["3+", "2", "1"],
    default="0"
)

survey_wide["buyer"] = survey_wide["buyer"].map({
    1: "yes",
    0: "no"
})

survey_final = survey_wide[
    [
        "buyer",
        "pairs",
        "STRIDE",
        "KESTREL",
        "EASYWALK"
    ]
].reset_index()

print("Survey data rows:", len(raw))

print("Clean yes answers:", len(survey_clean))

print("Adult panel members:", len(survey_final))

print("Non-buyers:",
      (survey_final["buyer"] == "no").sum())

print("People with two pair counts:",
      len(two_pair_counts))

bad_records_table = pd.DataFrame(bad_records)

print("\nBad records:")
display(bad_records_table)

# table: KEYWORD, your count, layout SAMPLE

comparison = pd.DataFrame({
    "KEYWORD": valid_keywords,
    "Your count": [
        (survey_final["buyer"] == "yes").sum(),
        (survey_final["pairs"] == "1").sum(),
        (survey_final["pairs"] == "2").sum(),
        (survey_final["pairs"] == "3+").sum(),
        survey_final["STRIDE"].sum(),
        survey_final["KESTREL"].sum(),
        survey_final["EASYWALK"].sum()
    ]
})

comparison = comparison.merge(
    survey_layout[["KEYWORD", "SAMPLE"]],
    on="KEYWORD",
    how="left"
)

comparison["SAMPLE"] = pd.to_numeric(
    comparison["SAMPLE"],
    errors="coerce"
)

comparison["Difference"] = (
    comparison["Your count"] - comparison["SAMPLE"]
)

print("\nComparison with layout SAMPLE:")
display(comparison)

raw_kestrel = (
    (raw["KEYWORD"] == "X710202")
    & (raw["VALUE"] == "1")
).sum()

clean_kestrel = survey_final["KESTREL"].sum()

raw_buyers = (
    (raw["KEYWORD"] == "X710100")
    & (raw["VALUE"] == "1")
).sum()

clean_buyers = (
    survey_final["buyer"] == "yes"
).sum()

raw_share = raw_buyers / len(adult_panel) * 100
clean_share = clean_buyers / len(adult_panel) * 100

kestrel_difference = raw_kestrel - clean_kestrel

kestrel_error = (
    kestrel_difference / clean_kestrel * 100
    if clean_kestrel != 0 else np.nan
)

share_difference = raw_share - clean_share

print("\nRaw vs clean data:")

print("Raw Kestrel count:", raw_kestrel)
print("Clean Kestrel count:", clean_kestrel)
print("Kestrel count difference:", kestrel_difference)
print("Kestrel relative error (%):", round(kestrel_error, 2))

print("Raw buyers:", raw_buyers)
print("Clean buyers:", clean_buyers)

print("Raw buyer share (%):", round(raw_share, 2))
print("Clean buyer share (%):", round(clean_share, 2))

print("Buyer share difference (percentage points):",
      round(share_difference, 2))

print("\nData validation:")

print("Duplicate respondents:",
      survey_final["RESPID"].duplicated().sum())

print("Buyers without pairs:",
      ((survey_final["buyer"] == "yes") &
       (survey_final["pairs"] == "0")).sum())

print("Non-buyers with pairs:",
      ((survey_final["buyer"] == "no") &
       (survey_final["pairs"] != "0")).sum())

print("Non-buyers with brands:",
      ((survey_final["buyer"] == "no") &
       (survey_final[["STRIDE", "KESTREL", "EASYWALK"]]
        .sum(axis=1) > 0)).sum())

print("\nFinal clean survey:")
display(survey_final.head(20))
```

```python
# ## Task 1.1 — Four definitions
# %%
person_weights = (
    daily_weights[daily_weights["resp_type"] == "R"]
    .groupby("respondent_id")["weight"]
    .mean()
    .reset_index()
)

person_weights["respondent_id"] = (
    person_weights["respondent_id"].astype(str).str.lstrip("0")
)

ages = viewer_attributes[
    viewer_attributes["attr_id"] == 300
][["respondent_id", "value"]].copy()

ages["respondent_id"] = (
    ages["respondent_id"].astype(str).str.lstrip("0")
)

ages["age"] = pd.to_numeric(ages["value"], errors="coerce")

ages = ages[["respondent_id", "age"]].drop_duplicates("respondent_id")

audience = survey_final.merge(
    person_weights,
    left_on="RESPID",
    right_on="respondent_id",
    how="left"
)

audience = audience.merge(
    ages,
    on="respondent_id",
    how="left"
)

adults = audience[audience["age"] >= 18]

shoe_buyers = adults[adults["buyer"] == "yes"]

heavy_buyers = shoe_buyers[shoe_buyers["pairs"] == "3+"]

kestrel_buyers = shoe_buyers[shoe_buyers["KESTREL"] == 1]

adults_25_54 = adults[adults["age"].between(25, 54)]

print("Adults 18+ (M):", round(adults["weight"].sum() / 1e6, 3))
print("Shoe buyers (M):", round(shoe_buyers["weight"].sum() / 1e6, 3))
print("Heavy buyers, 3+ pairs (M):", round(heavy_buyers["weight"].sum() / 1e6, 3))
print("Kestrel buyers (M):", round(kestrel_buyers["weight"].sum() / 1e6, 3))
print("Adults 25-54 (M):", round(adults_25_54["weight"].sum() / 1e6, 3))
print("Kestrel buyers in panel (n):", kestrel_buyers["RESPID"].nunique())

```

```python

# ## Task 1.2 — Who are they?
# %%
sex = viewer_attributes[
    viewer_attributes["attr_id"].isin([201, 202])
][["respondent_id", "attr_id"]].copy()

sex["respondent_id"] = (
    sex["respondent_id"].astype(str).str.lstrip("0")
)

sex["sex"] = sex["attr_id"].map({
    201: "Men",
    202: "Women"
})

sex = sex[["respondent_id", "sex"]].drop_duplicates("respondent_id")

audience = audience.drop(columns=["sex"], errors="ignore")

audience = audience.merge(
    sex,
    on="respondent_id",
    how="left"
)

audience["age_group"] = pd.cut(
    audience["age"],
    bins=[18, 35, 55, np.inf],
    labels=["18-34", "35-54", "55+"],
    right=False
)

audience["group"] = (
    audience["sex"] + " " +
    audience["age_group"].astype(str)
)

adults = audience[audience["age"] >= 18]

audiences = {
    "Adults 18+": adults,

    "Shoe buyers": adults[
        adults["buyer"] == "yes"
    ],

    "Heavy buyers": adults[
        (adults["buyer"] == "yes") &
        (adults["pairs"] == "3+")
    ],

    "Kestrel buyers": adults[
        adults["KESTREL"] == 1
    ],

    "Adults 25-54": adults[
        adults["age"].between(25, 54)
    ]
}

groups = [
    "Men 18-34",
    "Women 18-34",
    "Men 35-54",
    "Women 35-54",
    "Men 55+",
    "Women 55+"
]

adult_shares = (
    adults.groupby("group")["weight"].sum()
    / adults["weight"].sum()
)

results = []

for name, data in audiences.items():

    total_weight = data["weight"].sum()

    for group in groups:

        group_weight = data.loc[
            data["group"] == group,
            "weight"
        ].sum()

        share = (
            group_weight / total_weight
            if total_weight > 0 else np.nan
        )

        adult_share = adult_shares.get(group, 0)

        index = (
            share / adult_share * 100
            if adult_share > 0 else np.nan
        )

        results.append({
            "Audience": name,
            "Group": group,
            "Share (%)": round(share * 100, 2),
            "Index": round(index, 2)
        })

audience_profile = pd.DataFrame(results)

display(audience_profile)

share_table = audience_profile.pivot(
    index="Audience",
    columns="Group",
    values="Share (%)"
)


# table: audience x group, share and index vs adults

# **Answer:** which definition for Kestrel; why shoe buyers look like adults; one limit of fused data.
"""
1Heavy buyers
2Distribution to athletic shoe buyers is similar to general adult population
3It cannot establish whether watching particular tv show has stimulated user to purchase a product
"""
```

```python

# ## Task 2.1 — Programs by audience
# %%
selected_programs = [
    ("NBC", "Sunday Night Football"),
    ("ABC", "Monday Night Football"),
    ("ABC", "Dancing with the Stars"),
    ("CBS", "NCIS"),
    ("Fox News Channel", "Hannity"),
    ("HGTV", "Unsellable Houses"),
    ("Food Network", "Halloween Wars"),
    ("FX", "Grotesquerie")
]

p = programs.copy()
v = viewings.copy()
w = daily_weights.copy()
rates = rate_card.copy()

p["broadcast_date"] = pd.to_datetime(
    p["broadcast_date"]
).dt.normalize()

w["broadcast_date"] = pd.to_datetime(
    w["broadcast_date"]
).dt.normalize()

for col in ["start_ts", "end_ts"]:
    p[col] = pd.to_datetime(p[col])
    v[col] = pd.to_datetime(v[col])


v["respondent_id"] = (
    v["respondent_id"].astype(str).str.lstrip("0")
)

w["respondent_id"] = (
    w["respondent_id"].astype(str).str.lstrip("0")
)

adult_ids = set(
    audience.loc[audience["age"] >= 18, "RESPID"]
)

target_ids = set(
    audience.loc[
        (audience["age"] >= 18) &
        (audience["buyer"] == "yes") &
        (audience["pairs"] == "3+"),
        "RESPID"
    ]
)

w = w[
    (w["broadcast_date"].dt.month == 10) &
    (w["resp_type"] == "R") &
    (w["respondent_id"].isin(adult_ids))
].copy()

w["is_target"] = w["respondent_id"].isin(target_ids)

adult_universe_daily = (
    w.groupby("broadcast_date")["weight"].sum()
)

target_universe_daily = (
    w[w["is_target"]]
    .groupby("broadcast_date")["weight"]
    .sum()
    .reindex(adult_universe_daily.index, fill_value=0)
)

adult_universe = adult_universe_daily.mean()
target_universe = target_universe_daily.mean()

p = p[
    (p["broadcast_date"].dt.month == 10) &
    (p["daypart"] == "Prime")
].copy()

v = v[
    (v["resp_type"] == "R") &
    (v["respondent_id"].isin(adult_ids))
].copy()

rates["cost_30s_usd"] = pd.to_numeric(
    rates["cost_30s_usd"],
    errors="coerce"
)

results = []
telecast_results = []

for network, program in selected_programs:

    program_mask = (
        p["program_name"].str.strip().str.casefold()
        == program.casefold()
    )

    if program in [
        "Sunday Night Football",
        "Monday Night Football"
    ]:

        weekday = (
            6 if program == "Sunday Night Football" else 0
        )

        program_mask = (
            (program_mask |
             (p["program_name"].str.strip().str.casefold()
              == "nfl football")) &
            (p["broadcast_date"].dt.dayofweek == weekday)
        )

    telecasts = p[
        (p["network_name"] == network) &
        program_mask
    ]

    if telecasts.empty:
        print("Program not found:", network, program)
        continue

    program_results = []

    for _, telecast in telecasts.iterrows():

        start = telecast["start_ts"]
        end = telecast["end_ts"]
        date = telecast["broadcast_date"]

        duration = (end - start).total_seconds()

        if duration <= 0:
            continue

        ua = adult_universe_daily.get(date, np.nan)
        ut = target_universe_daily.get(date, np.nan)

        if not (ua > 0 and ut > 0):
            continue

        sessions = v[
            (v["network_code"] == telecast["network_code"]) &
            (v["start_ts"] < end) &
            (v["end_ts"] > start)
        ].copy()

        sessions["overlap_start"] = (
            sessions["start_ts"].clip(lower=start)
        )

        sessions["overlap_end"] = (
            sessions["end_ts"].clip(upper=end)
        )

        sessions["overlap"] = (
            sessions["overlap_end"] -
            sessions["overlap_start"]
        ).dt.total_seconds().clip(lower=0)

        watched = (
            sessions.groupby("respondent_id")["overlap"]
            .sum()
            .reset_index()
        )

        watched["overlap"] = watched["overlap"].clip(
            upper=duration
        )

        day_weights = w.loc[
            w["broadcast_date"] == date,
            ["respondent_id", "weight", "is_target"]
        ]

        watched = watched.merge(
            day_weights,
            on="respondent_id",
            how="inner"
        )

        watched["weighted_viewers"] = (
            watched["weight"] *
            watched["overlap"] / duration
        )

        adult_viewers = watched["weighted_viewers"].sum()

        target_viewers = watched.loc[
            watched["is_target"],
            "weighted_viewers"
        ].sum()

        adult_rating = adult_viewers / ua * 100
        target_rating = target_viewers / ut * 100

        network_rates = rates[
            rates["network_code"] == telecast["network_code"]
        ].copy()

        rate_names = network_rates["program_name"].fillna(
            ""
        ).str.strip().str.casefold()

        program_rates = network_rates[
            rate_names.isin([
                program.casefold(),
                telecast["program_name"].strip().casefold()
            ])
        ]

        if program_rates.empty:

            program_rates = network_rates[
                network_rates["daypart"]
                .str.strip()
                .str.casefold()
                .eq("prime") &
                rate_names.eq("")
            ]

        cost = program_rates["cost_30s_usd"].mean()

        program_results.append({
            "telecast_id": telecast["telecast_id"],
            "Program": program,
            "Network": network,
            "Date": date,
            "Adult viewers": adult_viewers,
            "Target viewers": target_viewers,
            "Adults rating": adult_rating,
            "Target rating": target_rating,
            "Cost per :30 ($)": cost
        })

    if not program_results:
        print("No valid telecasts:", program)
        continue

    telecast_results.extend(program_results)

    stats = pd.DataFrame(program_results)

    adult_viewers = stats["Adult viewers"].mean()
    target_viewers = stats["Target viewers"].mean()

    adult_rating = stats["Adults rating"].mean()
    target_rating = stats["Target rating"].mean()

    cost = stats["Cost per :30 ($)"].mean()

    composition = (
        target_viewers / adult_viewers * 100
        if adult_viewers > 0 else np.nan
    )

    index = (
        target_rating / adult_rating * 100
        if adult_rating > 0 else np.nan
    )

    adult_cpm = (
        cost / adult_viewers * 1000
        if adult_viewers > 0 else np.nan
    )

    target_cpm = (
        cost / target_viewers * 1000
        if target_viewers > 0 else np.nan
    )

    results.append({
        "Program": program,
        "Network": network,
        "Adults rating": adult_rating,
        "Target rating": target_rating,
        "Target viewers (K)": target_viewers / 1000,
        "Composition (%)": composition,
        "Index": index,
        "Cost per :30 ($)": cost,
        "Adults CPM ($)": adult_cpm,
        "Target CPM ($)": target_cpm
    })


program_audience = pd.DataFrame(results)

telecast_audience = pd.DataFrame(telecast_results)

display(program_audience.round(3))

# table: 8 programs, the columns of Task 2.1
print("Target universe (M):", round(target_universe / 1e6, 3))
```

```python

# ## Task 5.1 — The plan
# %%
budget = 1_000_000

plan = program_audience[[
    "Program",
    "Cost per :30 ($)",
    "Target viewers (K)",
    "Target CPM ($)"
]].dropna().copy()

plan = plan[
    (plan["Cost per :30 ($)"] > 0) &
    (plan["Target viewers (K)"] > 0)
].reset_index(drop=True)

costs = plan["Cost per :30 ($)"].to_numpy()

impressions = plan["Target viewers (K)"].to_numpy() * 1000

n = len(plan)

max_spots = 10

c = -impressions

constraints = LinearConstraint(
    costs.reshape(1, -1),
    lb=0,
    ub=budget
)

bounds = Bounds(
    lb=np.zeros(n),
    ub=np.full(n, max_spots)
)

integrality = np.ones(n)

result = milp(
    c=c,
    constraints=constraints,
    bounds=bounds,
    integrality=integrality
)

if not result.success:
    raise RuntimeError(result.message)

plan["Spots"] = np.rint(result.x).astype(int)

plan["Cost ($)"] = (
    plan["Spots"] * plan["Cost per :30 ($)"]
)

plan["Target impressions"] = (
    plan["Spots"] *
    plan["Target viewers (K)"] * 1000
)

plan = plan[plan["Spots"] > 0].copy()

plan_cost = plan["Cost ($)"].sum()

plan_impressions = plan["Target impressions"].sum()

remaining_budget = budget - plan_cost

plan_cpm = plan_cost / plan_impressions * 1000

display(plan[[
    "Program",
    "Spots",
    "Cost ($)",
    "Target impressions"
]].round(2))

# table: program, spots, cost, target impressions
print("Plan cost ($):", round(plan_cost, 2))
print("Plan target impressions (M):", round(plan_impressions / 1e6, 3))


# ## Task 5.2 — Recommendation
# **Answer:** 6-8 sentences.
"""
I recommend allocating Kestrel’s advertising budget to Dancing with the Stars, Unsellable Houses, Halloween Wars, and Grotesquerie
The plan prioritizes cost efficiency and target audience volume rather than the highest index alone
It achieves an estimated target CPM of $92
However, the heavy-buyer forecast has a MAPE of 71%
Approximately $826 thousand is attributable to adult viewers who are not heavy buyers
After the campaign, I would measure actual target impressions, reach, frequency, and target CPM
and compare delivery with forecast
"""


# ## Bonus
# %%
print("Kaggle username:", ...)

# **Answer:** what your model does, in 2-3 sentences.
```
